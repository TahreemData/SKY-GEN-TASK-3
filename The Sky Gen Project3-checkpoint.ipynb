```python
#  onthly sale By Category
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Order Date into datetime format
df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

# Keep rows with valid dates
valid_df = df.dropna(subset=["Order Date"]).copy()

# Create month column
valid_df["Month"] = (
    valid_df["Order Date"].dt.to_period("M").astype(str)
)

# Calculate monthly sales for each category
monthly_sales = valid_df.pivot_table(
    index="Month",
    columns="Category",
    values="Sales",
    aggfunc="sum"
).fillna(0)

# Create area graph
monthly_sales.plot(
    kind="area",
    figsize=(12, 6),
    stacked=False,
    alpha=0.5
)

plt.title("Monthly Sales by Category")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.legend(title="Category")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from pathlib import Path

# Create an output folder
output_folder = Path("Superstore_EDA_Outputs")
output_folder.mkdir(exist_ok=True)

# Save cleaned dataset
df.to_csv(output_folder / "Superstore_Cleaned.csv", index=False)

# Overall summary
summary = {
    "Total Records": len(df),
    "Total Columns": len(df.columns),
    "Duplicate Rows": df.duplicated().sum(),
    "Total Sales": df["Sales"].sum(),
    "Unique Orders": df["Order ID"].nunique(),
    "Unique Customers": df["Customer ID"].nunique()
}

if "Profit" in df.columns:
    summary["Total Profit"] = df["Profit"].sum()

if "Quantity" in df.columns:
    summary["Total Quantity"] = df["Quantity"].sum()

if "Discount" in df.columns:
    summary["Average Discount"] = df["Discount"].mean()

if "Profit" in df.columns and df["Sales"].sum() != 0:
    summary["Profit Margin (%)"] = (
        df["Profit"].sum() / df["Sales"].sum()
    ) * 100

summary_df = pd.DataFrame(
    list(summary.items()),
    columns=["Metric", "Value"]
)

print(summary_df)

summary_df.to_csv(
    output_folder / "EDA_Summary.csv",
    index=False
)
```

```python
# Category performance
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Number_of_Records=("Sales", "count")
)

if "Profit" in df.columns:
    category_analysis["Total_Profit"] = (
        df.groupby("Category")["Profit"].sum()
    )

category_analysis = category_analysis.sort_values(
    "Total_Sales", ascending=False
)

print("Category Analysis:")
print(category_analysis.round(2))

category_analysis.to_csv(
    output_folder / "Category_Analysis.csv"
)

# Regional performance
region_analysis = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Number_of_Orders=("Order ID", "nunique")
)

if "Profit" in df.columns:
    region_analysis["Total_Profit"] = (
        df.groupby("Region")["Profit"].sum()
    )

print("\nRegional Analysis:")
print(region_analysis.round(2))

region_analysis.to_csv(
    output_folder / "Region_Analysis.csv"
)
```
